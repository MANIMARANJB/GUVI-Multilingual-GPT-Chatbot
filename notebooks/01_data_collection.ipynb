```python
import requests
import pandas as pd

from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse
```

### Discover internal GUVI links.

```python
links = []

for tag in soup.find_all("a", href=True):
    full_url = urljoin(url, tag["href"])
    links.append(full_url)

print("Total links found:", len(links))
```

```python
# remove duplicates
links = list(set(links))

print("Unique links:", len(links))
```

```python
# keep only GUVI links
guvi_links = []

for link in links:
    if "guvi.in" in urlparse(link).netloc:
        guvi_links.append(link)

print("GUVI internal links:", len(guvi_links))
```

### Clean the URLs

```python
from urllib.parse import urljoin, urlparse, urlunparse
```

```python
def clean_url(url):
    parsed = urlparse(url)

    clean = parsed._replace(
        query="",
        fragment=""
    )

    return urlunparse(clean)
```

```python
cleaned_links = []

for link in guvi_links:
    link = clean_url(link)

    if link not in cleaned_links:
        cleaned_links.append(link)

print("Cleaned links:", len(cleaned_links))
```

### Exclude unwanted pages

```python
exclude_keywords = [
    "session/sso",
    "privacy-policy",
    "terms-of-service"
]
```

```python
filtered_links = []

for link in cleaned_links:

    if not any(
        keyword in link.lower()
        for keyword in exclude_keywords
    ):
        filtered_links.append(link)

print("Useful links:", len(filtered_links))
```

### Categorize pages

```python
def get_category(url):

    url = url.lower()

    if "/courses/" in url:
        return "Course"

    elif "/zen-class" in url:
        return "Live Program"

    elif "/faq" in url:
        return "FAQ"

    elif "/free-resources" in url:
        return "Resource"

    elif "/community/" in url:
        return "Community"

    elif "/code-kata" in url:
        return "Practice"

    elif "/webkata" in url:
        return "Practice"

    elif "/sqlkata" in url:
        return "Practice"

    elif "/skill-assessment" in url:
        return "Assessment"

    elif "/refund-policy" in url:
        return "Policy"

    elif "/about" in url:
        return "About"

    elif "/contact" in url:
        return "Contact"

    else:
        return "Other"
```

### Create the page scraping function

```python
def scrape_page(url):
    try:
        response = requests.get(
            url,
            timeout=10
        )

        if response.status_code != 200:
            return None

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        # Remove unnecessary HTML elements
        for tag in soup([
            "script",
            "style",
            "noscript"
        ]):
            tag.decompose()

        title = (
            soup.title.get_text(strip=True)
            if soup.title
            else ""
        )

        content = soup.get_text(
            separator=" ",
            strip=True
        )

        return {
            "url": url,
            "title": title,
            "category": get_category(url),
            "content": content
        }

    except requests.RequestException as error:
        print(f"Error scraping {url}: {error}")
        return None
```

```python
# Create a simple cleaning function

import re

def clean_text(text):
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # Remove repeated spaces around punctuation
    text = re.sub(r"\s+([.,!?;:])", r"\1", text)

    # Remove leading/trailing spaces
    text = text.strip()

    return text
```

```python
sample_text = test_df.iloc[0]["content"]

cleaned_sample = clean_text(sample_text)

print(cleaned_sample[:2000])
```

### Remove exact repeated sentences/fragments

```python
def remove_duplicate_segments(text):
    segments = re.split(r"(?<=[.!?])\s+", text)

    unique_segments = []
    seen = set()

    for segment in segments:
        segment = segment.strip()

        if segment and segment not in seen:
            unique_segments.append(segment)
            seen.add(segment)

    return " ".join(unique_segments)
```

### Create the final scraping function

```python
def scrape_guvi_page(url):
    try:
        response = requests.get(
            url,
            timeout=10
        )

        if response.status_code != 200:
            print(f"Skipped: {url} - Status {response.status_code}")
            return None

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        # Remove unnecessary HTML elements
        for tag in soup(["script", "style", "noscript"]):
            tag.decompose()

        # Extract page title
        title = (
            soup.title.get_text(strip=True)
            if soup.title
            else ""
        )

        # Extract page text
        content = soup.get_text(
            separator=" ",
            strip=True
        )

        # Clean the text
        content = clean_text(content)

        # Ignore pages with very little content
        if len(content) < 200:
            return None

        return {
            "url": url,
            "title": title,
            "category": get_category(url),
            "content": content
        }

    except requests.RequestException as error:
        print(f"Error: {url}")
        print(error)
        return None
```
